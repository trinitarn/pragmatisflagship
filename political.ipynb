```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("opponent.xlsx")

# Hitung jumlah 0 dan 1
sentiment_counts = (
    df["positif or negatif"]
    .value_counts()
    .reindex([0, 1], fill_value=0)
)

# Labels untuk pie
labels = ["−", "+"]

# Warna: merah negatif, hijau positif
colors = ["#F44336", "#4CAF50"]

# Plot pie chart
plt.figure()

plt.pie(
    sentiment_counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)

plt.title("Sentiment Distribution (+ vs −)")

plt.axis("equal")

plt.show()
```

```python
# so i have a vision of making a visualization in which I make the predictive score as the x axis, so it has to be from -1 to 1, and then the y ordinat is the frequency of reply, so we want to see which sentiment score got better engagement
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("opponent.xlsx")

plt.figure()

plt.scatter(
    df["predicted_score"],
    df["total_reply"]
)

plt.title("Sentiment Score vs Reply Engagement")

plt.xlabel("Sentiment Score (-1 to 1)")
plt.ylabel("Reply Count")

plt.show()
```

```python
import numpy as np

# Bikin interval sentiment
bins = np.linspace(-1, 1, 21)

df["sentiment_bin"] = pd.cut(
    df["predicted_score"],
    bins
)
reply_avg = df.groupby(
    "sentiment_bin"
)["total_reply"].mean()
plt.figure()

reply_avg.plot(
    kind="bar"
)

plt.title("Average Replies by Sentiment Score")

plt.xlabel("Sentiment Score Range")
plt.ylabel("Average Reply Count")

plt.xticks(rotation=90)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("opponent.xlsx")

# Cross-tabulation
cross = pd.crosstab(
    df["mention_opponent"],
    df["positif or negatif"]
)

# Rename columns (biar jelas)
cross.columns = ["Negative", "Positive"]

# Rename index
cross.index = ["No Opponent Mention", "Opponent Mention"]

# Plot stacked bar
cross.plot(
    kind="bar",
    stacked=True
)

plt.title("Sentiment of Comments Mentioning Opponents")

plt.xlabel("Opponent Mention")
plt.ylabel("Number of Comments")

plt.xticks(rotation=0)

plt.legend(title="Sentiment")

plt.show()
```

```python
from wordcloud import WordCloud
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("opponent.xlsx")

# =========================
# STOPWORDS
# =========================

stopwords = {
    "dan","yang","di","ke","dari",
    "ini","itu","aja","nya",
    "untuk","dengan","karena",
    "atau","pada","jadi",
    "adalah","seperti","tidak",
    "ya","ga","gak",
    "yg","dr","rt",
    "nih","dong","sih"
}

# Optional — hapus nama kandidat kalau terlalu dominan
stopwords.update({
    "anis",
    "ganjar",
    "prabowo"
})

# =========================
# FILTER TEXT
# =========================

text_o1_pos = " ".join(
    df[
        (df["mention_opponent"] == 1) &
        (df["predicted_score"] > 0)
    ]["clean_comment"].astype(str)
)

text_o1_neg = " ".join(
    df[
        (df["mention_opponent"] == 1) &
        (df["predicted_score"] < 0)
    ]["clean_comment"].astype(str)
)

text_o0_pos = " ".join(
    df[
        (df["mention_opponent"] == 0) &
        (df["predicted_score"] > 0)
    ]["clean_comment"].astype(str)
)

text_o0_neg = " ".join(
    df[
        (df["mention_opponent"] == 0) &
        (df["predicted_score"] < 0)
    ]["clean_comment"].astype(str)
)

# =========================
# PLOT 4 WORDCLOUD
# =========================

plt.figure(figsize=(14,10))

def make_wc(text, position, title):
    if text.strip():
        wc = WordCloud(
            width=800,
            height=400,
            background_color="white",
            stopwords=stopwords
        ).generate(text)

        plt.subplot(2,2,position)
        plt.imshow(wc)
        plt.axis("off")
        plt.title(title)

# Row 1 → opponent = 1
make_wc(text_o1_pos, 1, "Opponent = 1 | Positive")
make_wc(text_o1_neg, 2, "Opponent = 1 | Negative")

# Row 2 → opponent = 0
make_wc(text_o0_pos, 3, "Opponent = 0 | Positive")
make_wc(text_o0_neg, 4, "Opponent = 0 | Negative")

plt.tight_layout()
plt.show()
```
